## Parameters

````python
import pandas as pd

DATA_URL = "https://storage.googleapis.com/simmer-python-course-data/simmer-store/v1/"
````

## Read the file

````python
sessions = pd.read_csv(DATA_URL + "sessions_export_2024_02.csv")
sessions.shape
````

## Keep the referral rows

````python
referral = sessions[sessions["session_medium"] == "referral"]
len(referral)
````

## Total their sessions

````python
referral_sessions = referral["sessions"].sum()
referral_sessions
````
