## Parameters

````python
import pandas as pd

FILE_URL = "https://storage.googleapis.com/simmer-python-course-data/simmer-store/v1/sessions_export_2024_05.csv"
````

````python
may = pd.read_csv(FILE_URL, parse_dates=["date"])
````

## 1. How big is it?

````python
n_rows, n_columns = may.shape
may.tail()
````

## 2. Which dates does it cover?

````python
last_date = may["date"].max()
last_date
````

## 3. What's missing?

````python
may.info()
missing_campaigns = may["session_campaign"].isna().sum()
missing_campaigns
````

## 4. What are the ranges?

````python
max_revenue = may["revenue"].max()
may.describe()
````

## 5. What's in the categories?

````python
top_medium = may["session_medium"].value_counts().index[0]
may["session_medium"].value_counts()
````
